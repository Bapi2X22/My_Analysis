```python
import numpy as np
import matplotlib.pyplot as plt
import awkward as ak
from coffea.nanoevents import NanoEventsFactory, NanoAODSchema
import ROOT
import importlib.util
```

```python
file = "root://xrootd-cms.infn.it///store/mc/RunIII2024Summer24NanoAODv15/TTG-1Jets_TuneCP5_13p6TeV_amcatnloFXFXold-pythia8/NANOAODSIM/150X_mcRun3_2024_realistic_v2-v2/120000/c1987957-7aa3-4faa-81f4-981a40791a2f.root"
factory = NanoEventsFactory.from_root(
    f"{file}:Events",
    schemaclass=NanoAODSchema,
)
events = factory.events()
```

```python
path = "BTV/lib/btag_sf.py"

spec = importlib.util.spec_from_file_location("btag_sf", path)
btag_sf = importlib.util.module_from_spec(spec)
spec.loader.exec_module(btag_sf)

compute_btagging_efficiency = btag_sf.compute_btagging_efficiency_multifile
```

```python
Jets = events.Jet
```

```python
cs = compute_btagging_efficiency(Jets, ["btagUParTAK4B"], {"M": 0.1272})
```

```python
print(cs.keys())
print(list(cs.keys()))
```

```python
corr = cs["UParTAK4_efficiency"]
print(corr)
print("Description:", corr.description)

for inp in corr.inputs:
    print(inp.name, inp.type)
```

```python
eff = corr.evaluate("M", 5, 1.2, 100.0)
print(eff)
```
